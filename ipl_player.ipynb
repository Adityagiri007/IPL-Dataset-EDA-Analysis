```python
team_runs_city = merged_df.groupby(['city', 'batting_team'])['total_runs'].sum().reset_index()
team_runs_city = team_runs_city.sort_values(by='total_runs', ascending=False)
team_runs_city
```

```python
# Filter for a specific city (e.g., Mumbai)
city_data = team_runs_city[team_runs_city['city'] == 'Mumbai']

# Set team name as index for plotting
city_data.set_index('batting_team', inplace=True)

# Plot pie chart
city_data['total_runs'].plot(kind='pie', autopct='%1.1f%%', figsize=(8, 8), startangle=140)
plt.title('Team Runs Distribution in Mumbai')
plt.ylabel('')
plt.show()
```

```python

# Pivot for heatmap
heatmap_data = team_runs_city.pivot(index='batting_team', columns='city', values='total_runs').fillna(0)

# Plot heatmap
plt.figure(figsize=(14, 8))
sns.heatmap(heatmap_data, cmap='YlGnBu', annot=False)
plt.title('Team Runs per City')
plt.xlabel('City')
plt.ylabel('Team')
plt.show()
```

```python
batsman_city_avg = merged_df.groupby(['batter', 'city'])['batsman_runs'].mean().reset_index()

# Optional: Sort to see top performers
batsman_city_avg = batsman_city_avg.sort_values(by='batsman_runs', ascending=False)

batsman_city_avg.head(10)  # Top 10 average scores
```

```python
# Total runs per batter per city
total_runs = merged_df.groupby(['batter', 'city'])['batsman_runs'].sum()

# Matches (distinct match IDs) per batter per city
matches_played = merged_df.groupby(['batter', 'city'])['match_id'].nunique()

# Calculate average runs per match
batsman_avg_match = (total_runs / matches_played).reset_index(name='avg_runs_per_match')

# Optional: sort
batsman_avg_match = batsman_avg_match.sort_values(by='avg_runs_per_match', ascending=False)

city_runner=batsman_avg_match.head(10)
city_runner
```
